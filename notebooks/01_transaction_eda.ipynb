**Design Decisions**

Translating transaction-level EDA outputs into graph-design decisions:

- 887k directed account pairs out of 9.5m total transactions
    - Aggregate transactions by sender-receiver pair
- 88.07% of directed pairs repeat
    - Use transaction count as the first edge weight
- No exact duplicates
    - No deduplication step is required
- 13 currencies and strongly skewed amounts
    - Should not be summing raw amounts across currencies
- 6507 multi-location accounts
    - Keep one node per account
- 0.1039% laundering transactions
    - Use ranking metrics
- 17 laundering typologies
    - Report results separately by typology
- Final date contains only 8.4k transactions
    - Exclude the incomplete date from test evaluation
